```python
# Basic Libraries
import numpy as np
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt # we only need pyplot
sb.set() # set the default Seaborn style for graphics
```

```python
total = pd.read_csv('train.csv')
print(total)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split


# Split the Dataset into Train and Test
saleprice = pd.DataFrame(total['SalePrice'])   # Response
grlivarea = pd.DataFrame(total['GrLivArea'])   # Predictor

x_train, x_test, y_train, y_test = train_test_split(grlivarea, saleprice, test_size = 0.2, shuffle=True,random_state=104)

# Linear Regression using Train Data
linreg = LinearRegression()         # create the linear regression object
linreg.fit(x_train, y_train)        # train the linear regression model

# Coefficients of the Linear Regression line
print('Intercept of Regression \t: b = ', linreg.intercept_)
print('Coefficients of Regression \t: a = ', linreg.coef_)
print()

# Predict Total values corresponding to Predictors
y_train_pred = linreg.predict(x_train)

# Plot the Predictions
f = plt.figure(figsize=(16, 8))
plt.scatter(x_train, y_train, color = "green")
plt.scatter(x_train, y_train_pred, color = "red")
plt.show()
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

# Check the Goodness of Fit (on Train Data)
print("Goodness of Fit of Model \tTrain Dataset")
print("Explained Variance (R^2) \t:", linreg.score(x_train, y_train))
print("Mean Squared Error (MSE) \t:", mean_squared_error(y_train, y_train_pred))
print()

y_test_pred = linreg.predict(x_test)

# Check the Goodness of Fit (on Test Data)
print("Goodness of Fit of Model \tTest Dataset")
print("Explained Variance (R^2) \t:", linreg.score(x_test, y_test))
print("Mean Squared Error (MSE) \t:", mean_squared_error(y_test, y_test_pred))
print()
```

2) GrLivArea. The R^2 value on the test dataset is the closest to 1 and the MSE is also the lowest on the test dataset out of all 3 models, which means that the regression model is a good predictor for future values of the SalePrice, given GrLivArea.


```python
Q1G = total['GrLivArea'].quantile(0.25) 
Q3G = total['GrLivArea'].quantile(0.75)
IQRG = Q3G - Q1G
lower_boundg = Q1G - 1.5 * IQRG
upper_boundg = Q3G + 1.5 * IQRG
outliersg = total[(total['GrLivArea'] < lower_boundg) | (total['GrLivArea'] > upper_boundg)]

Q1S = total['SalePrice'].quantile(0.25) 
Q3S = total['SalePrice'].quantile(0.75)
IQRS = Q3S - Q1S
lower_bounds = Q1S - 1.5 * IQRS
upper_bounds = Q3S + 1.5 * IQRS
outlierss = total[(total['SalePrice'] < lower_bounds) | (total['SalePrice'] > upper_bounds)]

union_outliers = pd.concat([outliersg, outlierss])

new_data = pd.concat([union_outliers, total])
new_data = new_data.drop_duplicates(keep=False)

print(new_data)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

# Split the Dataset into Train and Test
new_saleprice = pd.DataFrame(new_data['SalePrice'])   # Response
new_grlivarea = pd.DataFrame(new_data['GrLivArea'])   # Predictor

print(new_saleprice)
print(new_grlivarea)

x_train, x_test, y_train, y_test = train_test_split(new_grlivarea, new_saleprice, test_size = 0.2, shuffle = True,random_state=104)

# Linear Regression using Train Data
linreg = LinearRegression()         # create the linear regression object
linreg.fit(x_train, y_train)        # train the linear regression model

# Coefficients of the Linear Regression line
print('Intercept of Regression \t: b = ', linreg.intercept_)
print('Coefficients of Regression \t: a = ', linreg.coef_)
print()

# Predict Total values corresponding to Predictors
y_train_pred = linreg.predict(x_train)

# Plot the Predictions
f = plt.figure(figsize=(16, 8))
plt.scatter(x_train, y_train, color = "green")
plt.scatter(x_train, y_train_pred, color = "red")
plt.show()
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

y_test_pred = linreg.predict(x_test)

# Check the Goodness of Fit (on Test Data)
print("Goodness of Fit of Model \tTest Dataset")
print("Explained Variance (R^2) \t:", linreg.score(x_test, y_test))
print("Mean Squared Error (MSE) \t:", mean_squared_error(y_test, y_test_pred))
print()
```
